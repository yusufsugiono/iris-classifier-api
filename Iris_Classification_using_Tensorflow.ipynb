# Simple Iris Classification Model using Tensorflow

## Data Preparation

### Import Library

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import joblib
import numpy as np
from tensorflow.keras.models import load_model
```

### Load Dataset

```python
iris = load_iris()
X = iris.data
y = iris.target
```

## Data Preprocessing

### Data Normalization

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

### One Hot Encoding

```python
y_encoded = to_categorical(y)
```

### Data Split

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_encoded, test_size=0.2, random_state=42)
```

## Modelling

### Build and Compile Model

```python
model = Sequential([
    Dense(10, activation='relu', input_shape=(4,)),
    Dense(8, activation='relu'),
    Dense(3, activation='softmax')
])

model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

### Training

```python
model.fit(X_train, y_train, epochs=50, batch_size=5, verbose=1, validation_split=0.1)
```

### Evaluation

```python
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"Test accuracy: {accuracy:.4f}")
```

### Export Model & Scaler

```python
model.save("iris_model.h5")
joblib.dump(scaler, "scaler_iris.pkl")
```

## Result

### Predict with New Data

```python
# Load model
model = load_model("iris_model.h5")

# Load scaler
scaler = joblib.load("scaler_iris.pkl")

# New input
new_input = np.array([[6.0,3.4,4.5,1.6,]])

# Transform dengan scaler
new_input_scaled = scaler.transform(new_input)

# Predict
prediction = model.predict(new_input_scaled)
predicted_class = np.argmax(prediction, axis=1)

class_names = ['setosa', 'versicolor', 'virginica']

print("Prediction result:", class_names[predicted_class[0]])
```
